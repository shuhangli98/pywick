```python
x = [3,4,10]
out = []
for i in range(2 ** len(x)):
    z = bin(i)[2:]
    z = '0'*(len(x) - len(z)) + z
    current_list = [x[y] for y, ybit in enumerate(z) if ybit == '1']
    out.append(current_list)
out
```

```python
x = [3,4,10,12]
out = []
old_y = [[]]
for _ in range(len(x)+1):
    new_y = []
    for y in old_y:
        for i in x:
            if y == [] or i > y[-1]:
                new_y.append(y + [i])
    out += old_y
    old_y = new_y
out
```
